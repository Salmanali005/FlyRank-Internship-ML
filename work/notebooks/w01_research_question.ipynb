## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")
print(df.columns.tolist())
```

```python
print("Total Pages:", len(df))

declining = df[df["trend_direction"] == "down"]
print("Declining pages:", len(declining))

declining_with_demand = declining[declining["impressions_90d"] >= 100]  
print("Declining pages with real demand:", len(declining_with_demand))
```

Out of 30,000 pages in the starter sample, all of them meet the basic bar for review (real impressions and at least 90 days old). Of these, 16,262 pages (54.2%) are declining, which is too many for a reviewer to check one by one. Most of these declining pages, 13,152 (80.9%), still get real search traffic, so the drop isn't just noise on pages nobody sees. This shows there are enough pages that genuinely need review, which is exactly what Lane 2 is meant to solve.
